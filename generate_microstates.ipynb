```python
def get_gfp(raw):
    gfp_data = extract_gfp_peaks(raw, min_peak_distance=3)
    return gfp_data

def csv_to_raw_array(csv_file_path, sfreq=250):
    """
    Convert a CSV file with EEG channel data into an mne.io.RawArray object.
    
    Parameters:
    csv_file_path (str): Path to the CSV file.
    sfreq (float): Sampling frequency of the data. Default is 250 Hz.
    
    Returns:
    raw (mne.io.RawArray): The raw EEG data as an MNE RawArray object.
    """
    # Read the CSV file into a DataFrame
    df = pd.read_csv(csv_file_path)

    # Split the space-separated values and convert to floats
    data = []
    for column in df.columns:
        # For each column, split the string values into floats
        channel_data = [list(map(float, value.split())) for value in df[column]]
        # Flatten the list of lists
        channel_data_flat = [item for sublist in channel_data for item in sublist]
        data.append(channel_data_flat)

    # Convert the data to a NumPy array and transpose it
    data = np.array(data)
    
    # Define the channel names (from your original header)
    channel_names = ['F7', 'F3', 'F4', 'F8', 'T3', 'C3', 'Cz', 'C4', 'T4', 'T5', 
                     'P3', 'Pz', 'P4', 'T6', 'O1', 'O2']
    
    # Define the channel types (assuming EEG for all)
    channel_types = ['eeg'] * len(channel_names)
    
    # Create the info object
    info = mne.create_info(ch_names=channel_names, sfreq=sfreq, ch_types=channel_types)
    
    # Create the RawArray object
    raw = mne.io.RawArray(data, info)
    raw.pick('eeg')
    raw.set_eeg_reference('average')
    ten_twenty_montage = mne.channels.make_standard_montage('standard_1020')
    raw.set_montage(ten_twenty_montage)
    
    return raw

for filename in os.listdir(r'..\database\csv\norm'):
    if filename.endswith(".csv"):
        csv_file_path = os.path.join(r'..\database\csv\norm', filename)
        raw = csv_to_raw_array(csv_file_path)
        gfp_data = get_gfp(raw)
        generate_microstates(gfp_data, f"{filename}", True)
```

```python
import os


def get_all_filenames(directory, read_mode="eeg"):
    # Ensure the directory path is correct
    directory_path = os.path.abspath(directory)

    # List all entries in the directory
    filenames = os.listdir(directory_path)

    # Optional: Get full paths
    if read_mode == "eeg":
        full_paths = [os.path.join(directory_path, filename) for filename in filenames if
                      "TXT" not in filename and "s" not in filename]

    else:
        full_paths = [os.path.join(directory_path, filename) for filename in filenames]

    return full_paths
```

```python
def read_file_return_raw(filename: str):
    raw = read_raw_edf(filename, preload=True)
    raw.pick('eeg')
    raw.set_eeg_reference('average')
    ten_twenty_montage = mne.channels.make_standard_montage('standard_1020')
    raw.set_montage(ten_twenty_montage)

    return raw
```

```python
from pycrostates.preprocessing import extract_gfp_peaks


def extract_gfp(raw):
    gfp_data = extract_gfp_peaks(raw, min_peak_distance=3)
    return gfp_data
```

```python
def define_epochs(raw):
    epochs = mne.make_fixed_length_epochs(raw, duration=2, preload=True)
    gfp_data = extract_gfp_peaks(epochs, min_peak_distance=3)
    return gfp_data
```

```python
from pycrostates.preprocessing import resample


# extract 1 resample of 100 random high GFP samples

def resample_gfp(gfp_data):
    resample = resample(gfp_data, n_resamples=1, n_samples=100)
    return resample[0]
```

```python
from pycrostates.cluster import ModKMeans


def generate_microstates(gfp_data, name: str):
    modK = ModKMeans(n_clusters=20, random_state=20)
    modK.fit(gfp_data, n_jobs=24, verbose="WARNING")
    if "h" not in name:
        modK.plot().savefig(f"{name.split('.')[0]}.png")
    else:
        modK.plot().savefig(f"{name.split('.')[0]}.png")
```

```python
def initiate_generation():
    directory = r"..\database\dataverse_files"
    file_names = get_all_filenames(directory)

    for name in file_names:
        raw = read_file_return_raw(name)
        gfp_data = extract_gfp(raw)
        # resampled = resample_gfp(gfp_data)
        generate_microstates(gfp_data, name)


initiate_generation()
```
